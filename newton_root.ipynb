```python
import sympy as sp

def newton_step(f, x_symbol, x_old):
    # 1. Calculate the derivative formula
    f_prime = sp.diff(f, x_symbol)
    
    # 2. Swap symbols for actual numbers
    # We substitute x_old into both the function and its derivative
    f_val = f.subs(x_symbol, x_old)
    slope_val = f_prime.subs(x_symbol, x_old)
    
    # 3. Apply the Newton-Raphson formula
    # x_new = x_old - f(x) / f'(x)
    x_new = x_old - (f_val / slope_val)
    
    return float(x_new) # Convert to a standard decimal number

# --- Main Setup ---

# Define the symbol
x = sp.symbols("x")

# Define the function: 5x^2 - 10
f = 5*x**2 - 10

# Initial guess
current_guess = 10

# Run the function once
next_guess = newton_step(f, x, current_guess)

print(f"Starting guess: {current_guess}")
print(f"Next guess: {next_guess}")
```
