```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def lindist_pdf(x, limits = (0,1)):
    a = limits[0]
    b = limits[1]
    return (np.abs(b-a)/(b-a))* (x>=a)*(x<=b)
```

```python
def lindist_cdf(x, limits = (0,1)):
    a = limits[0]
    b = limits[1]
    return (np.abs(x-a)/(b-a))* (x>=a)*(x<=b)+(x>b)
```

```python
x = np.linspace(1,2,1000)
```

```python
low = 1.25
high = 1.35
```

```python
p = lindist_pdf(x, (low,high))
c= lindist_cdf(x, (low,high))
```

```python
def plt_pdf_cdf(x,p,c):
    plt.figure(figsize=(12,3))
    plt.subplot(1,2,1)
    plt.plot(x,p)
    plt.title('PDF')
    plt.subplot(1,2,2)
    plt.plot(x,c)
    plt.title('CDF')
    plt.show()
    plt.plot(x,p, color='red', linewidth=2)
    plt.plot(x,c, color='blue', linewidth=2)
    plt.legend(['PDF', 'CDF'], loc=2)
    plt.title('PDF Vs CDF')
    plt.show()
```

```python
plt_pdf_cdf(x,p,c)
```
