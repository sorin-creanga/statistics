```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
import random
print("ready")
```

```python
np.random.seed(10)
pop_1 = stats.poisson.rvs(loc = 18, mu = 35, size =150000) # rvs - random variable selector
pop_2 = stats.poisson.rvs(loc = 18, mu = 10, size =100000) # mu - poisson coefficient; 
pop_total = np.concatenate((pop_1,pop_2))
```

```python
fig, axes = plt.subplots (1,3,figsize = (10,5))
axes[0].hist(pop_1)
axes[1].hist(pop_2)
axes[2].hist(pop_total)
plt.tight_layout()
```

```python
# means
print(f"Pop 1 mean: {pop_1.mean()}")
print(f"Pop 2 mean: {pop_2.mean()}")
print(f"Popupulation whole mean: {pop_total.mean()}")
```

```python
# calcluating confidence intervals
import numpy


sample_size = 1000
sample = np.random.choice(a= pop_total, size = sample_size)
sample_mean = sample.mean()

z_critical = stats.norm.ppf(q = 0.975)
print(z_critical) # norm..ppf returns STANDARD DEV that corver the area up to 97.5% under the curve

pop_stdev = (numpy.cov(pop_total)**2)/(len(pop_total)**2)
true_pop_dev = pop_total.std()
print (true_pop_dev)

print(pop_stdev)

from math import sqrt
q = 0.975

margin_error = z_critical* (true_pop_dev/sqrt(sample_size))

print(margin_error)
```
